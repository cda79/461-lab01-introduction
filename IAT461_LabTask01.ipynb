### Part 3: Filtering ###

1. Show only the rows where the age is above 60.

2. Show rows where the fare is above 100 and the passenger was in 1st class.

3. Identify and isolate rows with missing values in the age column.

```python
# 1. Show only the rows where the age is above 60.
old_passengers = df.loc[df["Age"] > 60]
old_passengers.head()
```

```python
# 2. Show rows where the fare is above 100 and the passenger was in 1st class.
expensive_passengers = df.loc[(df["Fare"] > 100) & (df["Pclass"] == 1)]
expensive_passengers.head()
```

```python
# 3. Identify and isolate rows with missing values in the age column.
age_na = df[(df["Age"].isna() & df["Age"].isnull())]
age_na.head()
```

### Part 5: Arithmetic and Data Alignment ###

1. Create a new column that converts fares from USD to Euros.

2. Subtract the average age from every age value.

3. Subtract a custom array from two selected columns in the dataset.

```python
# 1. Create a new column that converts fares from USD to Euros.
eur_rate = .86018
df['Fare_EUR'] = df['Fare'] * eur_rate
print(df[['Fare', 'Fare_EUR']].head())

print()

# 2. Subtract the average age from every age value.
print('Age BEFORE subtraction:')
print(df[['Age']].head())
print()

print('Age AFTER subtraction:')
avg_age = df['Age'].mean()

df_age = df.copy()
df_age['Age'] = df['Age'] - avg_age
print(df_age[['Age']].head())

print()

# 3. Subtract a custom array from two selected columns in the dataset.
# each column has 891 values - generate a random numpy array of random integers going from 0-10 until n=891
custom_array =  np.random.randint(0, 10, size=891)
df_custom = pd.DataFrame(custom_array)

df_custom_array = df.copy()
df_custom_array[['Fare', 'Age']] = df[['Fare', 'Age']].sub(custom_array, axis=0)

print('Custom Array:')
print(df_custom.head())
print()
print('Fare and Age BEFORE custom subtraction:')
print(df[['Fare', 'Age']].head())
print()
print('Fare and Age AFTER custom subtraction:')
print(df_custom_array[['Fare', 'Age']].head())
```

### Part 6: Applying Functions ###

1. Create a new column with the number of characters in each name.

2. Calculate the difference between fare and age for each row.

3. Round all numerical values in the dataset to one decimal place.



```python
#1. Create a new column with the number of characters in each name.
df['Name_length'] = df['Name'].str.len()
print(df[['Name', 'Name_length']].head())

print()
#2. Calculate the difference between fare and age for each row.
df['fare_age_diff'] = df['Fare'] - df['Age']
print(df[['Fare', 'Age', 'fare_age_diff']].head())

print()

#3. Round all numerical values in the dataset to one decimal place.
df_onedecimal = df.round(1)
print('Round numerical values to 1 decimal place:')
print(df[['PassengerId','Survived','Pclass','Age','Fare']].head())
print(df_onedecimal[['PassengerId','Survived','Pclass','Age','Fare']].head())
```

### Part 7: Mapping and Transformation ###
1. Convert categorical values in the gender column to numeric codes.

2. Create a new column that categorizes age into groups (e.g., child, adult, senior).



```python
# 1. Convert categorical values in the gender column to numeric codes.
df["Sex_number"] = df["Sex"].replace({"male": "0", "female": "1"})
print(df[['Sex_number','Sex']].head())
print()

# 2. Create a new column that categorizes age into groups (e.g., child, adult, senior).
bins = [0, 12, 19, 64, 100]
labels = ['Child', 'Teenager', 'Adult', 'Senior']
df['Age_group'] = pd.cut(df['Age'], bins=bins, labels=labels)
print(df[['Age', 'Age_group']].head(30))
```
